## はじめに

今回はPythonの仮想環境のマネジメントツール`uv`の使い方をまとめておく。

- [uv](https://docs.astral.sh/uv/)
> Rustで書かれた、非常に高速なPythonパッケージおよびプロジェクトマネージャー

`uv`のウェブサイトにも記載されている通り非常に高速に動作するのが特徴とのこと


## `uv`のコマンド

### Python バージョン

- `uv python install`: Python のバージョンをインストールする
- `uv python list`: 利用可能な Python バージョンを表示する
- `uv python find`: インストール済みの Python バージョンを探す
- `uv python pin`: 現在のプロジェクトで使う Python バージョンを固定する
- `uv python uninstall`: Python のバージョンをアンインストールする

### スクリプト

- `uv run`: スクリプトを実行する
- `uv add --script`: スクリプトに依存パッケージを追加する
- `uv remove --script`: スクリプトから依存パッケージを削除する

### プロジェクト

- `uv init`: 新しい Python プロジェクトを作成する
- `uv add`: プロジェクトに依存パッケージを追加する
- `uv remove`: プロジェクトから依存パッケージを削除する
- `uv sync`: プロジェクトの依存関係を環境に同期する
- `uv lock`: ロックファイルを作成する
- `uv run`: プロジェクト環境内でコマンドを実行する
- `uv tree`: 依存関係ツリーを表示する
- `uv build`: プロジェクトを配布用アーカイブにビルドする
- `uv publish`: パッケージインデックスに公開する

### ツール

- `uvx / uv tool run`: 一時的な環境でツールを実行する
- `uv tool install`: ツールをユーザー全体にインストールする
- `uv tool uninstall`: ツールをアンインストールする
- `uv tool list`: インストール済みツールを一覧表示する
- `uv tool update-shell`: ツールの実行ファイルをシェルで使えるように設定を更新する

### pip インターフェース

- `uv venv`: 新しい仮想環境を作成する
- `uv pip install`: 現在の環境にパッケージをインストールする
- `uv pip show`: インストール済みパッケージの詳細を表示する
- `uv pip freeze`: インストール済みパッケージとそのバージョンを一覧表示する
- `uv pip check`: 現在の環境の依存関係に矛盾がないか確認する
- `uv pip list`: インストール済みパッケージを一覧表示する
- `uv pip uninstall`: パッケージをアンインストールする
- `uv pip tree`: 環境の依存関係ツリーを表示する
- `uv pip compile`: requirements をロックファイルに変換する
- `uv pip sync`: ロックファイルに合わせて環境を同期する

### ユーティリティ

- `uv cache clean`: キャッシュ項目を削除する
- `uv cache prune`: 古くなったキャッシュ項目を削除する
- `uv cache dir`: uv のキャッシュディレクトリの場所を表示する
- `uv tool dir`: uv のツールディレクトリの場所を表示する
- `uv python dir`: uv がインストールした Python バージョンの保存場所を表示する
- `uv self update`: uv 自身を最新版に更新する

## 仮想環境を作成

まずは作業するディレクトリに移動する。`uv`はディレクトリごとに仮想環境を作れる。今回はこのjupyter notebookがあるディレクトリ。

````python
!ls
````

初期化するコマンドは`uv init`。実行するとPythonバージョンやライブラリを管理するためのファイルが生成される。

````python
!uv init
````

````python
!tree -a -L 1
````

`.python-version`はPythonのバーションを管理するためのファイル。現状は`3.10`となっている。

````python
!cat .python-version
````

pythonのバーションを変更したければ、`uv python pin 3.11.1`とすればファイルの内容を書き換えることができる。

````python
!uv python pin 3.11.1
````

````python
!cat .python-version
````

`pyproject.toml`はPythonのライブラリを管理するためのファイル。`requires-python`はPythonバージョンの制限をかけることができる。

````python
!cat pyproject.toml
````

`PyTorch`をインストールしたければ`uv add torch`を実行すれば即インストールされる。

````python
!uv add torch
````

````python
!uv run python -c "import torch; print(torch.__version__)"
````

`pyproject.toml`にインストールしたライブラリが依存関係付きで保存される。

````python
!cat pyproject.toml
````

ライブラリをインストールすると同時に、`.venv`が作成され、仮想環境が作成される。

````python
!tree -a -L 1
````

ライブラリのバージョンを変更したければ、`uv add torch==2.9.0`と実行すればよい。

````python
!uv add torch==2.9.0
````

````python
!cat pyproject.toml
````

jupyter notebookを使用して、仮想環境が正しく動作しているか確認する。

````python
!uv add ipykernel
````

このPythonが動いている仮想環境のパスやさきほどインストールしたライブラリのバージョンを表示。内容も問題ない。

````python
import sys
import os
import torch

cwd = os.getcwd()
print("Virtualenv: ", os.path.relpath(sys.executable, cwd))
print(f"Python version: {sys.version}")
print(f"Torch version: {torch.__version__}")
````

````python
x = torch.rand(5, 3)
print(x)
````

## 仮想環境を再現

ここまでで作成した仮想環境を別のサーバーないし環境で再現する。再現のためにグラフニューラルネットワークの学習のために利用し始めたvast-aiでサーバーをレンタルして再現する。

まずは、SSH接続するための公開鍵を作って登録する。`ssh-keygen -t rsa -b 4096`を実行する。

```python
$ ssh-keygen -t rsa -b 4096
$ ls ~/.ssh/ | grep id_rsa
id_rsa 
id_rsa.pub
```

`id_rsa.pub`の内容を`cat ~/.ssh/id_rsa.pub | pbcopy`でコピーしてvast-aiの「Keys」のところに貼り付けておく。

<div align="center"><img src="./vast-key.png" width="700"></div>

レンタルしたサーバーの接続情報をもとにSSH接続することで、アクセスできる。vast-aiの接続情報をもとに、`port`と`ssh1`の部分を書き換える必要があるので忘れずに。

```python
-- Proxy
$ ssh -i ~/.ssh/id_rsa -p xxxxx root@ssh1.vast.ai -L 8080:localhost:8080
```




<div align="center"><img src="./vast-set.png" width="700"></div>

あとはGithubなりから仮想環境を作成したディレクトリをクローンする。その後に`uv sync`を実行して環境をシンクする。シンクすると、設定ファイルをもとに`.venv`が再現される。また、アクセスしたサーバーには`uv`は入ってないので、インストールのコードを実行する。

```python
$ curl -LsSf https://astral.sh/uv/install.sh | sh

$ ls -a
.  ..  .python-version  README.md  main.py  note_tips38.ipynb  pyproject.toml  uv.lock

$ uv sync
Resolved 64 packages in 0.96ms
Audited 54 packages in 1ms

$ ls -a
.  ..  .python-version  .venv  README.md  main.py  note_tips38.ipynb  pyproject.toml  uv.lock

$ uv run python main.py
```

出力をコピーし忘れたので、リモート接続したときに同じことをしているので、その画像を参考までに載せておく。
<div align="center"><img src="./vast-cl.png" width="700"></div>

エディタのリモート接続でもサーバーにアクセスできる。その場合は、`config`に接続情報を書いておく、


```python
$ cat ~/.ssh/config
Host vast-ai
  User root
  HostName ssh1.vast.ai
  Port xxxx
  IdentityFile ~/.ssh/id_rsa
```

サーバーにアクセスできたら、jupyter notebookのための「jupyter」や「python」の拡張をリモート接続先でもインストールする。さきほどの内容の通り、`uv sync`して仮想環境を再現すればjupyter notebookでも同じことができる。
