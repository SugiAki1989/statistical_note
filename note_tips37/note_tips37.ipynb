## 標本平均を微分する

標本平均の定義は下記の通り。

$$
\bar{x}=\frac{1}{n}\sum_{i=1}^{n} x_{i} = \frac{1}{n} (x_{1} + ... + x_{k} + ... + x_{n})
$$

興味本位で、平均を微分する。平均を微分するとは、あるデータ$x_k$を少し動かしたら、平均はどれだけ動くか？を見るということになる。つまり、

$$
\frac{\partial \bar{x}}{\partial x_k}
$$

を求める。丁寧に書くと、

$$
\frac{\partial \bar{x}}{\partial x_k} = \frac{\partial }{\partial x_k}\frac{1}{n} \sum_{i=1}^{n} x_{i} =  \frac{1}{n}\sum_{i=1}^{n} \frac{\partial x_i}{\partial x_k}
$$

となる。

$$
\frac{\partial x_i}{\partial x_k}
=
\begin{cases}
1 & i = k \\
0 & i \ne k
\end{cases}
$$

なので

$$
\frac{\partial \bar{x}}{\partial x_k} = \frac{1}{n} 1 = \frac{1}{n} 
$$

となる。これは、1つのデータを1だけ増やすと$\frac{1}{n}$分だけ増えるということ。

```python
import numpy as np

x = [2,4,6,8,10]
mean = np.mean(x)
mean
```

この数列に対して、平均を求めると6である。$x_3 = 7$に1つずらす。そうなると$\frac{1}{n}$なので、$\frac{1}{5}=0.2$だけ増加するはず。

```python
x = [2,4,7,8,10]
mean = np.mean(x)
mean
```

平均が6.2なので、思った通り増加している。1増やすと0.2増加するので、5増やせば1増加する。

```python
x = [2,4,11,8,10]
mean = np.mean(x)
mean
```

ベクトルで書くと

$$
\bar{x} = \frac{1}{n} \mathbf{1}^{T}\mathbf{x}
$$

であり、勾配を求めると、

$$
\nabla x = \frac{1}{n} (1,1,...,1) = (\frac{1}{n}, \frac{1}{n}, ...,\frac{1}{n})
$$

なので、平均はすべてのデータを均等に扱うことがわかる。そのため、100くらいずれた外れ値が存在しても、$n=5$であれば、$100*0.2=20$の影響をうけるが、$n=100$であれば、$100*0.01=1$の影響をうける。つまり、$\frac{1}{n}$分、引きずられる。また、

$$
w_i = \frac{1}{n}
$$

とおくと

$$
\begin{split}
\bar{x} 
&= \frac{1}{n}x_1 + ... + \frac{1}{n}x_k + ... + \frac{1}{n}x_n \\
&= w_1 x_1 + ... + w_k x_k + ... + w_n x_n \\
&= \sum_i  w_i x_i
\end{split}
$$

線形結合でかけることがわかる。平均はデータの線形結合であり、線形推定量といえる。さきほどの内容を思い出すと、データ1点が推定量に与える影響である影響関数は、すべての点が同じ影響なので、影響関数は一定。
